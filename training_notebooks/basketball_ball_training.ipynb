```python
# Imports Python’s built-in Operating System interface. Used here to read environment variables
import os
# Imports PyTorch, the primary deep learning framework. Used to configure GPU hardware behavior, manage CUDA memory (VRAM)
import torch
# Imports the Roboflow SDK client class. Used to authenticate with your Roboflow account via an API key
from roboflow import Roboflow
# Imports the core YOLO model class from the Ultralytics framework. Used to instantiate, load pretrained weights
from ultralytics import YOLO

# 1. Critical Windows Multi-processing Guard
if __name__ == "__main__":
    # 2. Maximize RTX 40-series Tensor Cores (TF32 speedup)
    torch.backends.cuda.matmul.allow_tf32 = True
    torch.backends.cudnn.allow_tf32 = True

    # 3. Clear any cached VRAM before starting
    torch.cuda.empty_cache()

    # 4. Securely fetch Roboflow API key and download dataset
    # (This defines the 'dataset' variable so the code below doesn't crash)
    api_key = os.getenv("ROBOFLOW_API_KEY", "CyHStWaMAHu3qgwEsHXO")
    rf = Roboflow(api_key=api_key)
    project = rf.workspace("workspace-5ujvu").project(
        "basketball-players-fy4c2-vfsuv"
    )
    dataset = project.version(3).download("yolov8")

    # 5. Load the extra-large model
    model = YOLO("yolov8x.pt")

    # 6. Train with strict 8GB VRAM / Windows limits
    results = model.train(
        data=f"{dataset.location}/data.yaml",  # Path to dataset configuration file specifying train/val paths and class labels
        epochs=250,   # Total training iterations through the dataset
        imgsz=640,   # Input image resolution scaled to 640x640 pixels
        batch=8,  # Safe limit for yolov8x on 8GB VRAM to prevent crash/pagefile lag
        device=0,  # RTX 4060 Ti
        workers=2,  # Prevents 'DataLoader worker exited unexpectedly' on Windows
        amp=True,  # Mandatory: Uses mixed-precision FP16 to fit 8GB VRAM
        cache="ram",  # Loads dataset to system RAM, keeping SSD idle
        deterministic=False,  # Allows non-deterministic hardware speed optimizations
        close_mosaic=10,  # Disables mosaic augmentation at the end for accuracy
    )
```
